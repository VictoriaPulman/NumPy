Посчитаем среднюю скорость такси

```python
speed_arr =data[:,7] / (data[:,8]/3600) # расстояния из 8 столбца/ время в пути из 9 столбца переведенное часы
speed_arr.mean()
```

Посчитаем количество поездок в феврале

```python
data[data[:,1] == 2].shape[0] # вывести все поездки в феврале где 1 это столбец 2 и 2 это февраль, [0] это вывод строк в shape, без нее было б (13333,15)
```

Посчитаем количество поездок с чаевыми более 50 долларов

```python
data[data[:,-3] >50].shape[0]
```

```python
data[~np.isnan(np.where(data[:,-3] > 50, data[:,-3], np.nan))] # ~ это отрицание
```
